```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Upload the dataset from your computer
uploaded = files.upload()

# Read the uploaded CSV file
filename = next(iter(uploaded))
df = pd.read_csv(filename)

# Display the dataset
print("Dataset loaded successfully!")
print("\nFirst 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)
```

```python
# Display the number of rows and columns
print("Dataset Shape:", df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display data types and non-null counts
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

```

```python
# Select the score columns
scores = ["math score", "reading score", "writing score"]

# Calculate average marks
print("Average Scores:")
print(df[scores].mean())

# Calculate highest marks
print("\nHighest Scores:")
print(df[scores].max())

# Calculate lowest marks
print("\nLowest Scores:")
print(df[scores].min())

# Calculate median marks
print("\nMedian Scores:")
print(df[scores].median())

```

```python
gender_analysis = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

print(gender_analysis.round(2))

```

```python
import matplotlib.pyplot as plt

subjects = ["math score", "reading score", "writing score"]
average_scores = df[subjects].mean()

plt.figure(figsize=(8, 5))
plt.bar(average_scores.index, average_scores.values)

plt.title("Average Student Scores by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Score")
plt.ylim(0, 100)

plt.show()

```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["math score"], bins=10, edgecolor="black")

plt.title("Distribution of Mathematics Scores")
plt.xlabel("Mathematics Score")
plt.ylabel("Number of Students")

plt.show()

```

```python
gender_scores = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

gender_scores.plot(kind="bar", figsize=(9, 5))

plt.title("Average Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.legend(title="Subject")

plt.tight_layout()
plt.show()

```

```python
score_columns = ["math score", "reading score", "writing score"]

correlation = df[score_columns].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Student Scores")

plt.tight_layout()
plt.show()

```

```python
# Calculate average scores
scores = ["math score", "reading score", "writing score"]
averages = df[scores].mean()

print("===== EDA FINDINGS =====")

print("\n1. Dataset Overview")
print("Total students:", len(df))
print("Total columns:", len(df.columns))

print("\n2. Average Scores")
print(averages.round(2))

print("\n3. Highest Average Score")
print(averages.idxmax(), ":", round(averages.max(), 2))

print("\n4. Lowest Average Score")
print(averages.idxmin(), ":", round(averages.min(), 2))

print("\n5. Missing Values")
print("Total missing values:", df.isnull().sum().sum())

print("\n6. Duplicate Records")
print("Total duplicate rows:", df.duplicated().sum())

print("\n7. Correlation Between Scores")
print(df[scores].corr().round(2))

```
